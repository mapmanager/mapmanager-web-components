# NicePool in a Python notebook

This mounts the existing TypeScript/Vue NicePool component through the thin `NicePoolAnyWidget` adapter. Python supplies records and synchronizes state; it does not reimplement NicePool analysis or plotting.

```python
from nicepool_anywidget import NicePoolAnyWidget

rows = [
    {"id": "a", "accept": "yes", "condition": "control", "x": -2, "y": 1},
    {"id": "b", "accept": "yes", "condition": "control", "x": -1, "y": 3},
    {"id": "c", "accept": "no", "condition": "treated", "x": 1, "y": 5},
    {"id": "d", "accept": "yes", "condition": "treated", "x": 2, "y": None},
    {"id": "e", "accept": "yes", "condition": "treated", "x": 3, "y": 9},
]
pool = NicePoolAnyWidget(rows, row_id_column="id", theme="light", height=720)
pool
```

Selections made in the GUI synchronize back to Python:

```python
pool.selection
```

Python can also drive the selection:

```python
pool.selection = {"primaryRowId": "e", "selectedRowIds": ["b", "e"]}
```
